```python
import json
def dataclean(filename):
    with open (filename, "r") as f :
        information = json.load(f)
    return information 

data = dataclean("store.json")
```

```python
#getting individual value 
def structure_data(data):
    text_to_num = {"one":1, "two": 2 , "three":3 , "four":4, "five":5}
    cleaned_data = []
    unique_user = set()
    
    for user in data:
        raw_rating = str(user['rating']).strip().lower()
        if raw_rating in text_to_num :
           raw_rating = text_to_num[raw_rating]
        user['rating'] = raw_rating 

        raw_age = user.get("age") #getfunction return none if there is no value 
        if(raw_age == None):
               user['age'] = None
        #dededuplication
     
        if(user['name'].strip() in unique_user):
         continue
        unique_user.add(user['name'])
        cleaned_data.append(user)
    print(cleaned_data)
    data = cleaned_data
    print(len(data))   
structure_data(data)
```
